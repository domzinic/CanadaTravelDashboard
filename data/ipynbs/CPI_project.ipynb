```python
import wbgapi as wb
import pandas as pd

countries = {
    "CAN": "Canada", "USA": "United States", "FRA": "France", "ESP": "Spain",
    "GRC": "Greece", "GBR": "United Kingdom", "CHE": "Switzerland",
    "TUR": "Turkey", "JPN": "Japan", "THA": "Thailand", "CHN": "China",
    "ZAF": "South Africa", "AUS": "Australia", "NZL": "New Zealand",
}

# Pull all available years (1960 onward)
df = wb.data.DataFrame(
    "FP.CPI.TOTL.ZG",
    economy=list(countries.keys()),
    time=range(1960, 2026),
    labels=False,
)

# Columns come back as "YR1960", "YR1961", ...; clean them up
df.columns = [int(c.replace("YR", "")) for c in df.columns]

# Reshape: years as rows, countries as columns
df = df.T
df.index.name = "year"
df = df.rename(columns=countries)

print(df.tail())
df.to_csv("inflation_14_countries.csv")
```
