```python
import wbgapi as wb
import pandas as pd

codes = ["CAN","USA","FRA","ESP","GRC","GBR","CHE","TUR","JPN","THA","CHN","ZAF","AUS","NZL"]

gdp = wb.data.DataFrame(
    "NY.GDP.PCAP.PP.KD",
    economy=codes,
    time=range(2014, 2025),
    labels=False,
)
gdp.columns = [int(c.replace("YR", "")) for c in gdp.columns]

# One value per country for clustering: 10-year average (or the latest year)
gdp_feature = gdp.mean(axis=1).rename("gdp_per_capita_ppp")
print(gdp_feature)

gdp.to_csv("gdp_per_capita_ppp_yearly.csv")
```
